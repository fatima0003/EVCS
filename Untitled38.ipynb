```python
%%time

import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)


# ============================================================
# 1. Features and target
# ============================================================

X = data.drop(columns=['Label'])
y = data['Label']


# ============================================================
# 2. Stratified 80/20 train-test split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


# ============================================================
# 3. Decision Tree
# ============================================================

dt = DecisionTreeClassifier(
    random_state=0,
    class_weight='balanced'
)


# ============================================================
# 4. 3-Fold Stratified Cross-Validation
# ============================================================

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

start = time.time()

cv_scores = cross_val_score(
    dt,
    X_train,
    y_train,
    cv=cv,
    scoring='f1_macro',
    n_jobs=-1
)

cv_time = time.time() - start


print("\n3-Fold Macro F1 Scores:", cv_scores)
print("Mean CV Macro F1:", np.mean(cv_scores))
print("CV Time:", cv_time, "seconds")


# ============================================================
# 5. Train on complete training set
# ============================================================

start = time.time()

dt.fit(X_train, y_train)

training_time = time.time() - start

print("\nTraining time:", training_time, "seconds")


# ============================================================
# 6. Training performance
# ============================================================

y_train_pred = dt.predict(X_train)

train_accuracy = accuracy_score(y_train, y_train_pred)
train_macro_f1 = f1_score(
    y_train,
    y_train_pred,
    average='macro'
)

print("\nTraining Accuracy:", train_accuracy)
print("Training Macro F1:", train_macro_f1)


# ============================================================
# 7. Test prediction
# ============================================================

start = time.time()

y_pred = dt.predict(X_test)

prediction_time = time.time() - start

print("\nPrediction time:", prediction_time, "seconds")
print(
    "Prediction time per sample:",
    prediction_time / len(X_test) * 1000,
    "ms"
)


# ============================================================
# 8. Test evaluation
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

weighted_precision = precision_score(
    y_test,
    y_pred,
    average='weighted',
    zero_division=0
)

weighted_recall = recall_score(
    y_test,
    y_pred,
    average='weighted',
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_pred,
    average='weighted',
    zero_division=0
)

macro_precision = precision_score(
    y_test,
    y_pred,
    average='macro',
    zero_division=0
)

macro_recall = recall_score(
    y_test,
    y_pred,
    average='macro',
    zero_division=0
)

macro_f1 = f1_score(
    y_test,
    y_pred,
    average='macro',
    zero_division=0
)


print("\n==============================")
print("DECISION TREE RESULTS")
print("==============================")

print("Accuracy:", accuracy)

print("\nWeighted metrics:")
print("Precision:", weighted_precision)
print("Recall:", weighted_recall)
print("F1:", weighted_f1)

print("\nMacro metrics:")
print("Precision:", macro_precision)
print("Recall:", macro_recall)
print("F1:", macro_f1)


# ============================================================
# 9. Classification report
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


# ============================================================
# 10. Confusion Matrix
# ============================================================

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    linewidths=0.5,
    linecolor='red'
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Decision Tree Confusion Matrix")

plt.show()


# ============================================================
# 11. Tree complexity
# ============================================================

print("\nTree depth:", dt.get_depth())
print("Number of leaves:", dt.get_n_leaves())
```

```python
#CREATING 3 IID CLIENTS
from sklearn.model_selection import StratifiedKFold

N_CLIENTS = 3
RANDOM_STATE = 42

skf = StratifiedKFold(
    n_splits=N_CLIENTS,
    shuffle=True,
    random_state=RANDOM_STATE
)

clients = []

for client_id, (_, client_idx) in enumerate(
    skf.split(X_train, y_train), start=1
):

    X_client = X_train.iloc[client_idx].copy()
    y_client = y_train.iloc[client_idx].copy()

    clients.append({
        "client_id": client_id,
        "X": X_client,
        "y": y_client
    })

print(f"Number of clients: {len(clients)}")

for client in clients:
    print(
        f"Client {client['client_id']}: "
        f"{client['X'].shape[0]} samples"
    )
```

```python
import pandas as pd

for client in clients:
    client_id = client["client_id"]
    y_client = client["y"]

    counts = y_client.value_counts().sort_index()
    percentages = y_client.value_counts(normalize=True).sort_index() * 100

    distribution = pd.DataFrame({
        "Count": counts,
        "Percentage": percentages.round(2)
    })

    print(f"\nClient {client_id}")
    print(distribution)
```

```python
from sklearn.ensemble import RandomForestClassifier

local_models = []

for client in clients:

    client_id = client["client_id"]
    X_client = client["X"]
    y_client = client["y"]

    model = RandomForestClassifier(
        n_estimators=100,
        random_state=0,
        class_weight='balanced',
        n_jobs=-1
    )

    model.fit(X_client, y_client)

    local_models.append(model)

    print(
        f"Client {client_id} trained "
        f"({len(X_client)} samples)"
    )
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

local_results = []

for i, model in enumerate(local_models, start=1):

    y_pred_local = model.predict(X_test)

    results = {
        "Client": i,
        "Accuracy": accuracy_score(y_test, y_pred_local),
        "Macro Precision": precision_score(
            y_test, y_pred_local,
            average='macro',
            zero_division=0
        ),
        "Macro Recall": recall_score(
            y_test, y_pred_local,
            average='macro',
            zero_division=0
        ),
        "Macro F1": f1_score(
            y_test, y_pred_local,
            average='macro',
            zero_division=0
        )
    }

    local_results.append(results)

local_results_df = pd.DataFrame(local_results)

print(local_results_df)
```

```python
import copy

# Copy the first fitted model so all required fitted attributes are retained
federated_rf = copy.deepcopy(local_models[0])

# Collect trees from all clients
combined_trees = []

for model in local_models:
    combined_trees.extend(model.estimators_)

# Replace the trees with all client trees
federated_rf.estimators_ = combined_trees
federated_rf.n_estimators = len(combined_trees)

print("Federated RF created")
print("Total number of trees:", len(federated_rf.estimators_))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

y_pred_federated = federated_rf.predict(X_test)

federated_accuracy = accuracy_score(
    y_test, y_pred_federated
)

federated_macro_precision = precision_score(
    y_test,
    y_pred_federated,
    average='macro',
    zero_division=0
)

federated_macro_recall = recall_score(
    y_test,
    y_pred_federated,
    average='macro',
    zero_division=0
)

federated_macro_f1 = f1_score(
    y_test,
    y_pred_federated,
    average='macro',
    zero_division=0
)

print("\n==============================")
print("FEDERATED RANDOM FOREST")
print("==============================")

print("Accuracy:", federated_accuracy)
print("Macro Precision:", federated_macro_precision)
print("Macro Recall:", federated_macro_recall)
print("Macro F1:", federated_macro_f1)
```

```python
# ============================================================
# FEDERATED XGBOOST - IID SOFT VOTING
# ============================================================

import time
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

N_CLASSES = 7
XGB_ROUNDS = 100
RANDOM_STATE = 0

print("=" * 60)
print("FEDERATED XGBOOST - IID")
print("=" * 60)

print("Number of clients:", len(clients))
print("Number of features:", X_train.shape[1])
print("Number of classes:", N_CLASSES)
print("Trees per client:", XGB_ROUNDS)
print("Aggregation: Uniform Soft Voting")
print()

# ------------------------------------------------------------
# XGBOOST PARAMETERS
# ------------------------------------------------------------

params = {
    "objective": "multi:softprob",
    "num_class": N_CLASSES,
    "eval_metric": "mlogloss",
    "eta": 0.1,
    "max_depth": 6,
    "min_child_weight": 1,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "reg_lambda": 1.0,
    "tree_method": "hist",
    "nthread": -1,
    "seed": RANDOM_STATE
}

# ------------------------------------------------------------
# TRAIN XGBOOST MODEL ON EACH CLIENT
# ------------------------------------------------------------

local_models = []

start_time = time.time()

for client in clients:

    client_id = client["client_id"]

    X_client = client["X"]
    y_client = client["y"]

    print(
        f"Training Client {client_id} "
        f"({len(X_client)} samples)..."
    )

    # Convert data to NumPy
    X_client_np = X_client.values.astype(np.float32)
    y_client_np = y_client.values.astype(np.int32)

    # Create XGBoost training matrix
    dtrain = xgb.DMatrix(
        X_client_np,
        label=y_client_np
    )

    # Train local model
    model = xgb.train(
        params,
        dtrain,
        num_boost_round=XGB_ROUNDS
    )

    local_models.append(model)

    print(
        f"Client {client_id} complete "
        f"-> {model.num_boosted_rounds()} trees"
    )

training_time = time.time() - start_time

print()
print("Total training time:", round(training_time, 2), "seconds")

# ------------------------------------------------------------
# FEDERATED AGGREGATION
# ------------------------------------------------------------

print()
print("=" * 60)
print("FEDERATED AGGREGATION")
print("=" * 60)

# Convert global test data
X_test_np = X_test.values.astype(np.float32)
y_test_np = y_test.values.astype(np.int32)

dtest = xgb.DMatrix(X_test_np)

# Store predictions from each client
client_probabilities = []

prediction_start = time.time()

for i, model in enumerate(local_models, start=1):

    probabilities = model.predict(dtest)

    client_probabilities.append(probabilities)

    print(
        f"Client {i} prediction shape:",
        probabilities.shape
    )

# Shape:
# (number of clients, number of test samples, number of classes)

client_probabilities = np.asarray(client_probabilities)

# ------------------------------------------------------------
# UNIFORM SOFT VOTING
# ------------------------------------------------------------

# Average the class probabilities from all clients

global_probabilities = np.mean(
    client_probabilities,
    axis=0
)

# Select class with highest probability

global_predictions = np.argmax(
    global_probabilities,
    axis=1
)

prediction_time = time.time() - prediction_start

# ------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test_np,
    global_predictions
)

macro_precision = precision_score(
    y_test_np,
    global_predictions,
    average="macro",
    zero_division=0
)

macro_recall = recall_score(
    y_test_np,
    global_predictions,
    average="macro",
    zero_division=0
)

macro_f1 = f1_score(
    y_test_np,
    global_predictions,
    average="macro",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test_np,
    global_predictions,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# FINAL RESULTS
# ------------------------------------------------------------

print()
print("=" * 60)
print("FEDERATED XGBOOST RESULTS - IID")
print("=" * 60)

print(f"Accuracy:          {accuracy:.6f}")
print(f"Macro Precision:   {macro_precision:.6f}")
print(f"Macro Recall:      {macro_recall:.6f}")
print(f"Macro F1:          {macro_f1:.6f}")
print(f"Weighted F1:       {weighted_f1:.6f}")

print()
print(f"Training time:     {training_time:.2f} seconds")
print(f"Prediction time:   {prediction_time:.4f} seconds")
print(
    f"Time per sample:   "
    f"{prediction_time / len(X_test):.8f} seconds"
)

# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print()
print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test_np,
        global_predictions,
        labels=np.arange(N_CLASSES),
        zero_division=0,
        digits=4
    )
)

# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

print()
print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

cm = confusion_matrix(
    y_test_np,
    global_predictions,
    labels=np.arange(N_CLASSES)
)

cm_df = pd.DataFrame(
    cm,
    index=[f"Class {i}" for i in range(N_CLASSES)],
    columns=[f"Class {i}" for i in range(N_CLASSES)]
)

display(cm_df)

# ------------------------------------------------------------
# SAVE RESULTS IN A DICTIONARY
# ------------------------------------------------------------

fed_xgb_iid_results = {
    "Model": "Federated XGBoost",
    "Setting": "IID",
    "Aggregation": "Uniform Soft Voting",
    "Clients": len(clients),
    "Trees per Client": XGB_ROUNDS,
    "Accuracy": accuracy,
    "Macro Precision": macro_precision,
    "Macro Recall": macro_recall,
    "Macro F1": macro_f1,
    "Weighted F1": weighted_f1,
    "Training Time": training_time,
    "Prediction Time": prediction_time
}

print()
print("=" * 60)
print("DONE")
print("=" * 60)
```

```python
# ============================================================
# RANDOM FOREST - NON-IID FEDERATED LEARNING
# Dirichlet Partition + Tree Pooling
# ============================================================

import copy
import time
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

N_CLIENTS = 3
N_CLASSES = 7
N_ESTIMATORS = 100
ALPHA = 0.5
RANDOM_STATE = 42

print("=" * 60)
print("RANDOM FOREST - NON-IID FEDERATED LEARNING")
print("=" * 60)

print(f"Clients:       {N_CLIENTS}")
print(f"Dirichlet α:   {ALPHA}")
print(f"Trees/client:  {N_ESTIMATORS}")
print("Aggregation:   Tree Pooling")
print()

# ------------------------------------------------------------
# CREATE DIRICHLET NON-IID PARTITION
# ------------------------------------------------------------

rng = np.random.default_rng(RANDOM_STATE)

y_train_array = np.asarray(y_train)

client_indices = [[] for _ in range(N_CLIENTS)]

for class_label in range(N_CLASSES):

    # Indices belonging to this class
    class_indices = np.where(
        y_train_array == class_label
    )[0]

    # Shuffle class indices
    rng.shuffle(class_indices)

    # Draw Dirichlet proportions
    proportions = rng.dirichlet(
        np.repeat(ALPHA, N_CLIENTS)
    )

    # Convert proportions to integer counts
    counts = (
        proportions * len(class_indices)
    ).astype(int)

    # Fix rounding so all samples are assigned
    difference = (
        len(class_indices) - counts.sum()
    )

    for i in range(difference):
        counts[i % N_CLIENTS] += 1

    # Split indices according to counts
    start = 0

    for client_id in range(N_CLIENTS):

        end = start + counts[client_id]

        client_indices[client_id].extend(
            class_indices[start:end]
        )

        start = end

# Shuffle each client's final indices
for i in range(N_CLIENTS):
    rng.shuffle(client_indices[i])

# ------------------------------------------------------------
# CREATE NON-IID CLIENTS
# ------------------------------------------------------------

non_iid_clients = []

for client_id in range(N_CLIENTS):

    indices = client_indices[client_id]

    X_client = X_train.iloc[indices].copy()
    y_client = y_train.iloc[indices].copy()

    non_iid_clients.append({
        "client_id": client_id + 1,
        "X": X_client,
        "y": y_client
    })

# ------------------------------------------------------------
# SHOW CLIENT DISTRIBUTIONS
# ------------------------------------------------------------

print("=" * 60)
print("NON-IID CLIENT DISTRIBUTIONS")
print("=" * 60)

for client in non_iid_clients:

    client_id = client["client_id"]
    y_client = client["y"]

    counts = (
        y_client
        .value_counts()
        .reindex(range(N_CLASSES), fill_value=0)
    )

    print()
    print(f"Client {client_id}")
    print(f"Total samples: {len(y_client)}")

    for class_label, count in counts.items():
        percentage = (
            count / len(y_client) * 100
        )

        print(
            f"  Class {class_label}: "
            f"{count:5d} ({percentage:6.2f}%)"
        )

# ------------------------------------------------------------
# TRAIN LOCAL RANDOM FORESTS
# ------------------------------------------------------------

local_rf_non_iid = []

local_results = []

print()
print("=" * 60)
print("TRAINING LOCAL RANDOM FORESTS")
print("=" * 60)

total_training_start = time.time()

for client in non_iid_clients:

    client_id = client["client_id"]

    X_client = client["X"]
    y_client = client["y"]

    print()
    print(
        f"Training Client {client_id} "
        f"({len(X_client)} samples)..."
    )

    model = RandomForestClassifier(
        n_estimators=N_ESTIMATORS,
        random_state=0,
        class_weight="balanced",
        n_jobs=-1
    )

    start = time.time()

    model.fit(
        X_client,
        y_client
    )

    training_time = time.time() - start

    local_rf_non_iid.append(model)

    # --------------------------------------------------------
    # EVALUATE LOCAL MODEL ON GLOBAL TEST SET
    # --------------------------------------------------------

    start = time.time()

    predictions = model.predict(X_test)

    prediction_time = time.time() - start

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    macro_precision = precision_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    macro_recall = recall_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    weighted_f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    local_results.append({
        "Client": client_id,
        "Samples": len(X_client),
        "Accuracy": accuracy,
        "Macro Precision": macro_precision,
        "Macro Recall": macro_recall,
        "Macro F1": macro_f1,
        "Weighted F1": weighted_f1,
        "Training Time": training_time,
        "Prediction Time": prediction_time
    })

    print(
        f"Accuracy: {accuracy:.6f} | "
        f"Macro F1: {macro_f1:.6f}"
    )

total_training_time = (
    time.time() - total_training_start
)

# ------------------------------------------------------------
# DISPLAY LOCAL RESULTS
# ------------------------------------------------------------

print()
print("=" * 60)
print("LOCAL NON-IID RANDOM FOREST RESULTS")
print("=" * 60)

local_rf_non_iid_df = pd.DataFrame(
    local_results
)

display(local_rf_non_iid_df)

# ------------------------------------------------------------
# FEDERATED TREE POOLING
# ------------------------------------------------------------

print()
print("=" * 60)
print("FEDERATED RANDOM FOREST AGGREGATION")
print("=" * 60)

# Start from a fitted RF so all required sklearn
# attributes are preserved.

federated_rf_non_iid = copy.deepcopy(
    local_rf_non_iid[0]
)

combined_trees = []

for model in local_rf_non_iid:

    combined_trees.extend(
        model.estimators_
    )

# Replace trees with all client trees
federated_rf_non_iid.estimators_ = (
    combined_trees
)

federated_rf_non_iid.n_estimators = (
    len(combined_trees)
)

print(
    f"Trees from Client 1: "
    f"{N_ESTIMATORS}"
)

print(
    f"Trees from Client 2: "
    f"{N_ESTIMATORS}"
)

print(
    f"Trees from Client 3: "
    f"{N_ESTIMATORS}"
)

print(
    f"Total federated trees: "
    f"{len(combined_trees)}"
)

# ------------------------------------------------------------
# EVALUATE FEDERATED RANDOM FOREST
# ------------------------------------------------------------

start = time.time()

fed_predictions = (
    federated_rf_non_iid.predict(X_test)
)

fed_prediction_time = time.time() - start

fed_accuracy = accuracy_score(
    y_test,
    fed_predictions
)

fed_macro_precision = precision_score(
    y_test,
    fed_predictions,
    average="macro",
    zero_division=0
)

fed_macro_recall = recall_score(
    y_test,
    fed_predictions,
    average="macro",
    zero_division=0
)

fed_macro_f1 = f1_score(
    y_test,
    fed_predictions,
    average="macro",
    zero_division=0
)

fed_weighted_f1 = f1_score(
    y_test,
    fed_predictions,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# FEDERATED RESULTS
# ------------------------------------------------------------

print()
print("=" * 60)
print("FEDERATED RANDOM FOREST - NON-IID RESULTS")
print("=" * 60)

print(f"Accuracy:          {fed_accuracy:.6f}")
print(f"Macro Precision:   {fed_macro_precision:.6f}")
print(f"Macro Recall:      {fed_macro_recall:.6f}")
print(f"Macro F1:          {fed_macro_f1:.6f}")
print(f"Weighted F1:       {fed_weighted_f1:.6f}")

print(
    f"\nPrediction Time:   "
    f"{fed_prediction_time:.4f} seconds"
)

print(
    f"Time/sample:       "
    f"{fed_prediction_time / len(X_test):.8f} seconds"
)

# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print()
print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        fed_predictions,
        labels=np.arange(N_CLASSES),
        zero_division=0,
        digits=4
    )
)

# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

print()
print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

cm = confusion_matrix(
    y_test,
    fed_predictions,
    labels=np.arange(N_CLASSES)
)

cm_df = pd.DataFrame(
    cm,
    index=[f"Class {i}" for i in range(N_CLASSES)],
    columns=[f"Class {i}" for i in range(N_CLASSES)]
)

display(cm_df)

# ------------------------------------------------------------
# SAVE FEDERATED RESULTS
# ------------------------------------------------------------

fed_rf_non_iid_results = {
    "Model": "Federated Random Forest",
    "Setting": "Non-IID",
    "Partition": "Dirichlet",
    "Alpha": ALPHA,
    "Clients": N_CLIENTS,
    "Trees per Client": N_ESTIMATORS,
    "Total Trees": len(combined_trees),
    "Accuracy": fed_accuracy,
    "Macro Precision": fed_macro_precision,
    "Macro Recall": fed_macro_recall,
    "Macro F1": fed_macro_f1,
    "Weighted F1": fed_weighted_f1,
    "Prediction Time": fed_prediction_time
}

print()
print("=" * 60)
print("DONE")
print("=" * 60)
```

```python
# ============================================================
# XGBOOST - NON-IID FEDERATED LEARNING
# Dirichlet Partition + Soft Voting
# ============================================================

import time
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

N_CLIENTS = 3
N_CLASSES = 7
N_ESTIMATORS = 100
ALPHA = 0.5
RANDOM_STATE = 42

print("=" * 60)
print("XGBOOST - NON-IID FEDERATED LEARNING")
print("=" * 60)

print(f"Clients:       {N_CLIENTS}")
print(f"Dirichlet α:   {ALPHA}")
print(f"Trees/client:  {N_ESTIMATORS}")
print("Aggregation:   Uniform Soft Voting")
print()

# ------------------------------------------------------------
# CREATE DIRICHLET NON-IID PARTITION
# ------------------------------------------------------------

rng = np.random.default_rng(RANDOM_STATE)

y_train_array = np.asarray(y_train)

client_indices = [[] for _ in range(N_CLIENTS)]

for class_label in range(N_CLASSES):

    class_indices = np.where(
        y_train_array == class_label
    )[0]

    rng.shuffle(class_indices)

    # Dirichlet distribution determines
    # how each class is distributed among clients
    proportions = rng.dirichlet(
        np.repeat(ALPHA, N_CLIENTS)
    )

    counts = (
        proportions * len(class_indices)
    ).astype(int)

    # Correct rounding so every sample is assigned
    difference = (
        len(class_indices) - counts.sum()
    )

    for i in range(difference):
        counts[i % N_CLIENTS] += 1

    start = 0

    for client_id in range(N_CLIENTS):

        end = start + counts[client_id]

        client_indices[client_id].extend(
            class_indices[start:end]
        )

        start = end

# Shuffle final client indices
for i in range(N_CLIENTS):
    rng.shuffle(client_indices[i])

# ------------------------------------------------------------
# CREATE NON-IID CLIENTS
# ------------------------------------------------------------

non_iid_xgb_clients = []

for client_id in range(N_CLIENTS):

    indices = client_indices[client_id]

    X_client = X_train.iloc[indices].copy()
    y_client = y_train.iloc[indices].copy()

    non_iid_xgb_clients.append({
        "client_id": client_id + 1,
        "X": X_client,
        "y": y_client
    })

# ------------------------------------------------------------
# SHOW CLIENT DISTRIBUTIONS
# ------------------------------------------------------------

print("=" * 60)
print("NON-IID CLIENT DISTRIBUTIONS")
print("=" * 60)

for client in non_iid_xgb_clients:

    client_id = client["client_id"]
    y_client = client["y"]

    counts = (
        y_client
        .value_counts()
        .reindex(range(N_CLASSES), fill_value=0)
    )

    print()
    print(f"Client {client_id}")
    print(f"Total samples: {len(y_client)}")

    for class_label, count in counts.items():

        percentage = (
            count / len(y_client) * 100
        )

        print(
            f"  Class {class_label}: "
            f"{count:5d} ({percentage:6.2f}%)"
        )

# ------------------------------------------------------------
# XGBOOST PARAMETERS
# ------------------------------------------------------------

params = {
    "objective": "multi:softprob",
    "num_class": N_CLASSES,
    "eval_metric": "mlogloss",

    "eta": 0.1,
    "max_depth": 6,
    "min_child_weight": 1,

    "subsample": 0.8,
    "colsample_bytree": 0.8,

    "reg_lambda": 1.0,

    "tree_method": "hist",
    "nthread": -1,

    "seed": 0
}

# ------------------------------------------------------------
# TRAIN LOCAL XGBOOST MODELS
# ------------------------------------------------------------

local_xgb_non_iid = []

local_results = []

print()
print("=" * 60)
print("TRAINING LOCAL XGBOOST MODELS")
print("=" * 60)

total_training_start = time.time()

for client in non_iid_xgb_clients:

    client_id = client["client_id"]

    X_client = client["X"]
    y_client = client["y"]

    print()
    print(
        f"Training Client {client_id} "
        f"({len(X_client)} samples)..."
    )

    # Convert to NumPy
    X_client_np = X_client.values.astype(np.float32)
    y_client_np = y_client.values.astype(np.int32)

    # Create DMatrix
    dtrain = xgb.DMatrix(
        X_client_np,
        label=y_client_np
    )

    start = time.time()

    model = xgb.train(
        params,
        dtrain,
        num_boost_round=N_ESTIMATORS
    )

    training_time = time.time() - start

    local_xgb_non_iid.append(model)

    print(
        f"Client {client_id} complete "
        f"-> {model.num_boosted_rounds()} trees"
    )

    # --------------------------------------------------------
    # LOCAL MODEL TEST PERFORMANCE
    # --------------------------------------------------------

    dtest = xgb.DMatrix(
        X_test.values.astype(np.float32)
    )

    start = time.time()

    probabilities = model.predict(dtest)

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    prediction_time = time.time() - start

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    macro_precision = precision_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    macro_recall = recall_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    weighted_f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    local_results.append({
        "Client": client_id,
        "Samples": len(X_client),
        "Accuracy": accuracy,
        "Macro Precision": macro_precision,
        "Macro Recall": macro_recall,
        "Macro F1": macro_f1,
        "Weighted F1": weighted_f1,
        "Training Time": training_time,
        "Prediction Time": prediction_time
    })

    print(
        f"Accuracy: {accuracy:.6f} | "
        f"Macro F1: {macro_f1:.6f}"
    )

total_training_time = (
    time.time() - total_training_start
)

# ------------------------------------------------------------
# LOCAL RESULTS TABLE
# ------------------------------------------------------------

print()
print("=" * 60)
print("LOCAL NON-IID XGBOOST RESULTS")
print("=" * 60)

local_xgb_non_iid_df = pd.DataFrame(
    local_results
)

display(local_xgb_non_iid_df)

# ------------------------------------------------------------
# FEDERATED SOFT-VOTING AGGREGATION
# ------------------------------------------------------------

print()
print("=" * 60)
print("FEDERATED XGBOOST AGGREGATION")
print("=" * 60)

dtest = xgb.DMatrix(
    X_test.values.astype(np.float32)
)

client_probabilities = []

start = time.time()

for i, model in enumerate(
    local_xgb_non_iid,
    start=1
):

    probabilities = model.predict(dtest)

    client_probabilities.append(
        probabilities
    )

    print(
        f"Client {i} probability shape: "
        f"{probabilities.shape}"
    )

# Shape:
# (clients, test samples, classes)

client_probabilities = np.asarray(
    client_probabilities
)

# ------------------------------------------------------------
# UNIFORM SOFT VOTING
# ------------------------------------------------------------

global_probabilities = np.mean(
    client_probabilities,
    axis=0
)

global_predictions = np.argmax(
    global_probabilities,
    axis=1
)

fed_prediction_time = (
    time.time() - start
)

# ------------------------------------------------------------
# FEDERATED METRICS
# ------------------------------------------------------------

fed_accuracy = accuracy_score(
    y_test,
    global_predictions
)

fed_macro_precision = precision_score(
    y_test,
    global_predictions,
    average="macro",
    zero_division=0
)

fed_macro_recall = recall_score(
    y_test,
    global_predictions,
    average="macro",
    zero_division=0
)

fed_macro_f1 = f1_score(
    y_test,
    global_predictions,
    average="macro",
    zero_division=0
)

fed_weighted_f1 = f1_score(
    y_test,
    global_predictions,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# FINAL RESULTS
# ------------------------------------------------------------

print()
print("=" * 60)
print("FEDERATED XGBOOST - NON-IID RESULTS")
print("=" * 60)

print(f"Accuracy:          {fed_accuracy:.6f}")
print(f"Macro Precision:   {fed_macro_precision:.6f}")
print(f"Macro Recall:      {fed_macro_recall:.6f}")
print(f"Macro F1:          {fed_macro_f1:.6f}")
print(f"Weighted F1:       {fed_weighted_f1:.6f}")

print(
    f"\nPrediction Time:   "
    f"{fed_prediction_time:.4f} seconds"
)

print(
    f"Time/sample:       "
    f"{fed_prediction_time / len(X_test):.8f} seconds"
)

# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print()
print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_test,
        global_predictions,
        labels=np.arange(N_CLASSES),
        zero_division=0,
        digits=4
    )
)

# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

print()
print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

cm = confusion_matrix(
    y_test,
    global_predictions,
    labels=np.arange(N_CLASSES)
)

cm_df = pd.DataFrame(
    cm,
    index=[f"Class {i}" for i in range(N_CLASSES)],
    columns=[f"Class {i}" for i in range(N_CLASSES)]
)

display(cm_df)

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

fed_xgb_non_iid_results = {
    "Model": "Federated XGBoost",
    "Setting": "Non-IID",
    "Partition": "Dirichlet",
    "Alpha": ALPHA,
    "Clients": N_CLIENTS,
    "Trees per Client": N_ESTIMATORS,
    "Aggregation": "Uniform Soft Voting",
    "Accuracy": fed_accuracy,
    "Macro Precision": fed_macro_precision,
    "Macro Recall": fed_macro_recall,
    "Macro F1": fed_macro_f1,
    "Weighted F1": fed_weighted_f1,
    "Prediction Time": fed_prediction_time
}

print()
print("=" * 60)
print("DONE")
print("=" * 60)
```
